```python
import pandas as pd
import requests

# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/life-expectancy.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
df.head(5)
```

```python
df.rename(columns={"entity": "País"}, inplace=True)
```

```python
df.head(3)
```

```python
df.rename(columns={"code": "Código",
                   "year": "Año",
                   "life_expectancy_0": "Esperanza"}, inplace=True)
```

```python
df.set_index("País",inplace=True)
```

```python
df.set_index("País","Año", inplace=True)
```

```python
df.local(df["País"].str.contains{"U", case=False, regrex=True})
```

```python
df.local(df["Código"].notnull{})
```

```python
df.dropna(subset=["Código"], inplace=True)
df.isnull().sum()
```

```python
resultado {
    df.groupby("País")["Esperanza"]
    .mean()
    .sort_values(ascending=True)
}
```
